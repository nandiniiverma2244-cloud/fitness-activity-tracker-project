```python
import datetime

class Workout:
    def __init__(self, activity_type, duration_minutes, intensity):
        self.date = datetime.date.today()
        self.activity_type = activity_type
        self.duration_minutes = duration_minutes
        self.intensity = intensity.lower() # 'low', 'moderate', 'high'
        self.calories_burned = self._calculate_calories()

    def _calculate_calories(self):
        # Rough estimate of MET (Metabolic Equivalent of Task) values
        met_values = {
            'running': {'low': 7, 'moderate': 9.8, 'high': 11.5},
            'cycling': {'low': 4, 'moderate': 7.5, 'high': 10},
            'swimming': {'low': 5, 'moderate': 7, 'high': 9},
            'walking': {'low': 3, 'moderate': 3.5, 'high': 4.5},
            'strength': {'low': 3, 'moderate': 5, 'high': 6}
        }

        activity = self.activity_type.lower()
        # Fallback to general cardio if activity not listed
        met_dict = met_values.get(activity, {'low': 4, 'moderate': 6, 'high': 8})
        met = met_dict.get(self.intensity, 6)

        # Standard calorie formula: (MET * 3.5 * weight_kg) / 200 * duration
        # Using an average weight of 70kg for simplicity
        weight_kg = 70
        calories = (met * 3.5 * weight_kg) / 200 * self.duration_minutes
        return round(calories)

class FitnessTracker:
    def __init__(self, user_name):
        self.user_name = user_name
        self.workout_history = []

    def log_workout(self, activity_type, duration_minutes, intensity):
        workout = Workout(activity_type, duration_minutes, intensity)
        self.workout_history.append(workout)
        print(f"Successfully logged {duration_minutes} mins of {activity_type} ({intensity} intensity). Estimate: {workout.calories_burned} kcal burned!")

    def display_summary(self):
        if not self.workout_history:
            print("No workouts logged yet!")
            return

        print(f"\n=== Fitness Summary for {self.user_name} ===")
        total_calories = 0
        total_duration = 0

        for i, workout in enumerate(self.workout_history, 1):
            print(f"{i}. {workout.date} | {workout.activity_type.capitalize()} | {workout.duration_minutes} mins | {workout.intensity} intensity | {workout.calories_burned} kcal")
            total_calories += workout.calories_burned
            total_duration += workout.duration_minutes

        print("-" * 40)
        print(f"Total workouts: {len(self.workout_history)}")
        print(f"Total time active: {total_duration} minutes")
        print(f"Total calories burned: {total_calories} kcal")
        print("=" * 40)

# Instantiate and test the fitness tracker
tracker = FitnessTracker("Alex")
tracker.log_workout("Running", 30, "moderate")
tracker.log_workout("Strength", 45, "high")
tracker.log_workout("Walking", 20, "low")

tracker.display_summary()
```
